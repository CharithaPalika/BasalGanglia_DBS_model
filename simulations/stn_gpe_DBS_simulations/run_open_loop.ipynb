```python
import os, sys

# Root project
project_root = os.path.abspath(os.path.join(os.getcwd(), "../.."))

if project_root not in sys.path:
    sys.path.append(project_root)

print("Project root added to sys.path:", project_root)


import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
from stn_gpe import run_open_loop_DBS
import os
```

# Amp = 0.2

```python
rng = np.random.default_rng(42)
mean = 1
std_dev = 0.02
N = 100
wk = rng.normal(loc=mean, scale=std_dev,size=N)

# Time Span
t_span = (0, 4000)
t_eval = np.arange(t_span[0], t_span[-1], 0.01) # total time = 40sec  
dt = t_span[-1]/np.size(t_eval) 
```

```python
k = 0.45
freq = 16
pulse_amp = 0.2
G_X, P_a, G_Y, G_Z, x_10 = run_open_loop_DBS(t_eval, wk,  k_val = k, freq = freq, P_a = pulse_amp)
```

## Saving data

```python
folder_name = 'temp/open_loop'
folder_path = os.path.join(project_root, folder_name)

# Create folder if it doesn't exist
os.makedirs(folder_path, exist_ok=True)

np.savetxt(os.path.join(folder_path, 'X_mean_amp_0.2.txt'), G_X, delimiter=",")
np.savetxt(os.path.join(folder_path, 'Y_mean_amp_0.2.txt'), G_Y, delimiter=",")
np.savetxt(os.path.join(folder_path, 'Z_mean_amp_0.2.txt'), G_Z, delimiter=",")
np.savetxt(os.path.join(folder_path, 'DBS_pulses_amp_0.2.txt'), P_a, delimiter=",")
np.savetxt(os.path.join(folder_path, 'x_osc_amp_0.2.txt'), x_10, delimiter=",")
```

## Plotting


```python
Pulses = np.loadtxt(os.path.join(folder_path, 'DBS_pulses_amp_0.2.txt'), delimiter=',')
X = np.loadtxt(os.path.join(folder_path, 'X_mean_amp_0.2.txt'), delimiter=',')

plt.figure(figsize = (12, 4))
plt.plot(t_eval, X, label = 'X', color = 'tab:blue')
plt.plot(t_eval, Pulses, label = 'DBS pulses', color = 'red')
plt.legend()
```
