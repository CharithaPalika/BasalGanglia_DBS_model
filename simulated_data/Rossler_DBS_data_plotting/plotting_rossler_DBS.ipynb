## Closed Loop

```python
dbs_pulses_closed = np.loadtxt(r'DATA/CLOSED_LOOP/K_0.45/DBS_pulses.csv', delimiter=',')
X_mean_closed = np.loadtxt(r'DATA/CLOSED_LOOP/K_0.45/X_mean.csv', delimiter=',')
# x_data_closed = np.loadtxt(r'DATA\CLOSED_LOOP\K_0.45\x_oscillators.csv', delimiter=',')
```

```python
sr = 12000 # saved data at a downsampled time samples to save memory so sr is different

# --- FFT Computation ---
fft_output_pulses = fft(dbs_pulses_closed)
fft_output_X_without_dbs = fft(X_mean_closed[0:50000])
fft_output_Xdbs = fft(X_mean_closed[150000:200000])

# Lengths
N_p = len(fft_output_pulses)
N_x = len(fft_output_X_without_dbs)
N_xdbs = len(fft_output_Xdbs)

# Frequency axes (positive half only)
freq_p_closed = np.fft.fftfreq(N_p, 1/sr)[:N_p // 2]
freq_x_closed = np.fft.fftfreq(N_x, 1/sr)[:N_x // 2]
freq_xdbs_closed = np.fft.fftfreq(N_xdbs, 1/sr)[:N_xdbs // 2]

# Raw FFT magnitudes (positive frequency only)
mag_pulses = np.abs(fft_output_pulses[:N_p // 2])
mag_X_wo_dbs = np.abs(fft_output_X_without_dbs[:N_x // 2])
mag_X_dbs = np.abs(fft_output_Xdbs[:N_xdbs // 2])

# --- Gaussian Smoothing ---
sigma = 5  # Adjust for more or less smoothing
mag_pulses_smooth_closed = gaussian_filter1d(mag_pulses, sigma=sigma)
mag_X_wo_dbs_smooth_closed = gaussian_filter1d(mag_X_wo_dbs, sigma=sigma)
mag_X_dbs_smooth_closed = gaussian_filter1d(mag_X_dbs, sigma=sigma)
```

```python
from scipy.signal import welch
from scipy.integrate import trapezoid

def bandpower(signal, sr, low, high, nperseg=None):
    """
    Band power via Welch's method (averaged, windowed periodogram) --
    NOT a raw FFT-magnitude sum, which is what the mag_* variables above
    are (amplitude spectrum from a single unwindowed FFT, no averaging).
    Returns power integrated between `low` and `high` Hz.
    """
    signal = np.asarray(signal)
    if nperseg is None:
        nperseg = min(len(signal), sr)   # ~1 Hz resolution, capped at segment length
    freqs, psd = welch(signal, fs=sr, nperseg=nperseg, noverlap=nperseg // 2)
    band_mask = (freqs >= low) & (freqs <= high)
    return trapezoid(psd[band_mask], freqs[band_mask])

beta_low, beta_high = 10, 50   # NOTE: wider than the conventional 13-30 Hz beta band --
                                 # confirm this matches your intended definition before publishing

# DBS-on windows are NOT equal length across conditions in this notebook (closed loop:
# 200000 samples / 16.7s; the other three: 100000 samples / 8.3s). Truncating all four
# to the shortest common window so the comparison across conditions is fair.
dbs_window_len = 50000

beta_power = {}
beta_power['closed_loop_baseline']    = bandpower(X_mean_closed[0:50000], sr, beta_low, beta_high)
beta_power['closed_loop_dbs']         = bandpower(X_mean_closed[350000:350000 + dbs_window_len], sr, beta_low, beta_high)
beta_power['standard_dbs_baseline']   = bandpower(X_mean_std[0:50000], sr, beta_low, beta_high)
beta_power['standard_dbs_dbs']        = bandpower(X_mean_std[350000:350000 + dbs_window_len], sr, beta_low, beta_high)
beta_power['open_loop_low_baseline']  = bandpower(X_mean_open_low[0:50000], sr, beta_low, beta_high)
beta_power['open_loop_low_dbs']       = bandpower(X_mean_open_low[350000:350000 + dbs_window_len], sr, beta_low, beta_high)
beta_power['open_loop_high_baseline'] = bandpower(X_mean_open_high[0:50000], sr, beta_low, beta_high)
beta_power['open_loop_high_dbs']      = bandpower(X_mean_open_high[350000:350000 + dbs_window_len], sr, beta_low, beta_high)

for k, v in beta_power.items():
    print(f"{k:28s}: {v:.4e}")

print()
for cond in ['closed_loop', 'standard_dbs', 'open_loop_low', 'open_loop_high']:
    baseline = beta_power[f'{cond}_baseline']
    dbs = beta_power[f'{cond}_dbs']
    pct_change = 100 * (dbs - baseline) / baseline
    print(f"{cond:16s}: {pct_change:+.1f}% beta power change with intervention")
```
