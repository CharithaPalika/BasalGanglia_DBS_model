## Closed-loop DBS

```python
folder_path = project_root + '/basal_ganglia/rossler_dbs_stn_data/closed_loop'

Pulses = np.loadtxt(os.path.join(folder_path, 'DBS_pulses_closedloop.txt'), delimiter=',')
X = np.loadtxt(os.path.join(folder_path, 'X_mean_closedloop.txt'), delimiter=',')
x_osc = np.loadtxt(os.path.join(folder_path, 'x_osc_closedloop.txt'), delimiter=',')
x_osc = x_osc[299999:399999, 0:4]
```

```python
osc_output = x_osc.reshape(-1, 100, x_osc.shape[1]).mean(axis=1)
normalized_x_vals = (osc_output - np.min(osc_output))/(np.max(osc_output)-np.min(osc_output))
```

```python
std = np.std(normalized_x_vals, axis=1)
mean = np.mean(normalized_x_vals, axis=1)

plt.plot(std)
```

```python
np.mean(std)
```
